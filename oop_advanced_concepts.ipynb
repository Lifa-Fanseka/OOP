<h2>Class Methods</h2>

<p>
A <strong>class method</strong> is a method that belongs to the class rather than to a specific object.
</p>

<p>
Class methods use the decorator <code>@classmethod</code> and take <code>cls</code> as the first parameter.
</p>

<p>
They are typically used when a method needs to work with class-level data or create objects in a controlled way.
</p>

```python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name

    @classmethod
    def get_college(cls):
        return cls.college
```

```python
print(Student.get_college())
```

<h2>Static Methods</h2>

<p>
A <strong>static method</strong> does not depend on instance data or class data.
</p>

<p>
It behaves like a regular function but is grouped inside a class because it is logically related.
</p>

<p>
Static methods use the decorator <code>@staticmethod</code>.
</p>

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
```

```python
print(Calculator.add(5, 3))
```

<h2>Composition vs Inheritance</h2>

<p>
In previous lectures, we used inheritance to reuse code. However, inheritance is not always the best choice.
</p>

<p>
Composition is often preferred because it creates more flexible and maintainable designs.
</p>

<p>
Inheritance represents an <strong>is-a</strong> relationship.
</p>

<p>
Composition represents a <strong>has-a</strong> relationship.
</p>

```python
# Inheritance example
class Engine:
    pass

class Car(Engine):   # Not logical
    pass
```

```python
# Composition example
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()
```

<h2>Why Composition Is Often Better</h2>

<ul>
<li>It reduces tight coupling.</li>
<li>It improves flexibility.</li>
<li>It avoids deep inheritance hierarchies.</li>
<li>It aligns better with real-world relationships.</li>
</ul>

<h2>Separation of Concerns (SoC)</h2>

<p>
Separation of Concerns means dividing a program into distinct sections, where each section has a specific role.
</p>

<p>
For example, in a student system:
</p>

<ul>
<li>A class should handle student data.</li>
<li>A separate class should handle calculations.</li>
<li>A separate class should handle reporting.</li>
</ul>

<p>
This prevents one class from becoming too complex and difficult to maintain.
</p>

<h2>Single Responsibility Principle (SRP)</h2>

<p>
The Single Responsibility Principle states that:
</p>

<blockquote>
<strong>A class should have one, and only one, reason to change.</strong>
</blockquote>

<p>
This means each class should perform one clear role.
</p>

<p>
For example:
</p>

<ul>
<li>A <code>Student</code> class should store student data.</li>
<li>A <code>Calculator</code> class should perform calculations.</li>
<li>A <code>ReportGenerator</code> class should handle output.</li>
</ul>

<h2>Bad Design Example</h2>

<p>
The following example violates SRP:
</p>

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_average(self):
        return sum(self.marks) / len(self.marks)

    def save_to_file(self):
        print("Saving to file...")

    def print_report(self):
        print("Printing report...")
```

<h2>Improved Design</h2>

<p>
We improve the design by separating responsibilities:
</p>

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


class Calculator:
    def average(self, marks):
        return sum(marks) / len(marks)


class ReportGenerator:
    def generate(self, student, avg):
        print(student.name, avg)
```

<h2>Introduction to SOLID Principles</h2>

<p>
SOLID is a set of five design principles that help developers create maintainable software.

<ul>
<li><strong>S</strong> – Single Responsibility Principle</li>

<li><strong>I</strong> – Interface Segregation Principle</li>
<li><strong>D</strong> – Dependency Inversion Principle</li>
</ul>

<h2>Real-World System Thinking</h2>

<p>
When designing systems, always ask:
