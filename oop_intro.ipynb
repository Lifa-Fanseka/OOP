<h2>Why Object-Oriented Programming Matters</h2>

<p>
As software systems grow in size and complexity, it becomes increasingly difficult to manage code using only
variables, functions, and procedural logic. Real-world systems usually need to model meaningful entities such as
students, employees, bank accounts, invoices, vehicles, bookings, and customers.
</p>

<p>
Object-oriented programming helps us organise code by combining related data and related behaviour into a single
unit called an object.
</p>

<p>
For example, consider a student in an academic system. A student has data such as:
</p>

<ul>
    <li>Name.</li>
    <li>Student number.</li>
    <li>Qualification.</li>
    <li>Marks.</li>
</ul>

<p>
A student may also perform actions such as:
</p>

<ul>
    <li>Calculating an average mark.</li>
    <li>Displaying details.</li>
    <li>Determining pass or fail status.</li>
</ul>

<p>
Instead of storing this information in disconnected variables and functions, OOP allows us to model it neatly
inside a <code>Student</code> object.
</p>

<blockquote>
<strong>A class describes what something is, and an object is one real example of that thing.</strong>
</blockquote>

<ul>

<h2>Classes and Objects</h2>

<p>
A <strong>class</strong> is a blueprint or template used to create objects. It defines the data that objects
should store and the actions they should perform.
</p>

<p>
An <strong>object</strong> is a specific instance created from a class.
</p>

<p>
For example, if we define a <code>Student</code> class, we may create many student objects:
</p>

<ul>
    <li>A student named Ayesha.</li>
    <li>A student named John.</li>
    <li>A student named Thabo.</li>
</ul>

<p>
Each object is created from the same class, but each object stores different data.
</p>

<p>
Python convention recommends using <strong>PascalCase</strong> for class names. This means each word begins
with a capital letter.
</p>

<ul>
    <li><code>Student</code></li>
    <li><code>BankAccount</code></li>
    <li><code>VehicleBooking</code></li>
</ul>

<h2>Creating a Simple Class</h2>

<p>
A class is created using the <code>class</code> keyword.
</p>

<pre><code>class Student:
    pass</code></pre>

<p>
The keyword <code>pass</code> tells Python that the block is intentionally empty.
This is useful when planning a program structure before adding details.
</p>

<p>
At this stage, the class does not yet store data or define behaviour, but objects can already be created from it.
</p>

```python
class Student:
    pass # This is an empty class definition.
```

<h2>Creating Objects from a Class</h2>

<p>
Once a class has been defined, we create objects by calling the class name as though it were a function.
</p>

<pre><code>s1 = Student()</code></pre>

<p>
This creates a new object from the <code>Student</code> class and stores a reference to it in
<code>s1</code>.
</p>

<p>
If we print the object directly, Python displays information about the object’s type and memory location.
This confirms that an object has been created successfully.
</p>

<p>
Variables such as <code>s1</code> do not usually store the object itself directly. Instead, they store a
reference to the object in memory.
</p>

```python
s1 = Student() # Create an instance of the Student class.
print(type(s1)) # This will print the memory address of the s1 object.
```

<h2>The Constructor: <code>__init__()</code></h2>

<p>
Most objects need starting values when they are created. For example, a student object should usually have
a name and student number immediately.
</p>

<p>
Python uses a special method called <code>__init__()</code> for this purpose.
</p>

<pre><code>class Student:

    def __init__(self, name, number):
        self.name = name
        self.number = number</code></pre>

<p>
When an object is created, Python automatically calls <code>__init__()</code>.
</p>

<pre><code>s1 = Student("John", "2026001")</code></pre>

<p>
The double underscores indicate that this is a special built-in method. These are often called
<strong>dunder methods</strong>, meaning “double underscore methods”.
</p>

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student(name={self.name}, age={self.age})" 
s1 = Student("Pete", 20)
print(s1)
```

```python
class Bus:
    def __init__(self, name, registration):
        self.name = name
        self.registration = registration
    def __str__(self):
        return f"The name of the bus driver is {self.name} and the registration number is {self.registration}"
driver1= Bus("John", 202020)
driver2= Bus("Lifa", 202021)

drivers=[driver1, driver2]
for driver in drivers:
    print(driver)    
```

```python
s1 = Student("Pete", 20)
print(s1) # This will print the string representation of the s1 object.
```

<h2>Understanding <code>self</code></h2>

<p>
One of the most important differences between Python and C# is the explicit use of <code>self</code>.
</p>

<p>
The word <code>self</code> refers to the current object. It allows an object to access its own data and methods.
</p>

<pre><code>self.name = name</code></pre>

<p>
On the left-hand side, <code>self.name</code> creates or updates an attribute belonging to the object.
On the right-hand side, <code>name</code> refers to the value passed into the constructor.
</p>

<p>
Although <code>self</code> appears as the first parameter in instance methods, we do not supply it manually.
Python supplies it automatically.
</p>

<pre><code>s1.display()</code></pre>

<p>
In this example, Python automatically passes <code>s1</code> as <code>self</code>.
</p>

<p>
Students familiar with C# can think of <code>self</code> as being similar to <code>this</code>.
</p>

<h2>Instance Attributes</h2>

<p>
An <strong>instance attribute</strong> is a variable that belongs to one specific object.
</p>

<p>
For example, two student objects may have different names and student numbers. These values must therefore
belong to each object individually.
</p>

<pre><code>self.name = name
self.student_number = student_number</code></pre>

<p>
Each object receives its own copy of these attributes.
Changing one student’s name does not affect another student object.
</p>

<h2>Adding Behaviour with Methods</h2>

<p>
Classes should not only store data. They should also define behaviour through methods.
</p>

<p>
A <strong>method</strong> is a function that belongs to a class and usually operates on the object’s data.
</p>

<pre><code>class Student:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, I am {self.name}.")</code></pre>

<p>
Usage:
</p>

<pre><code>s1 = Student("Mary")
s1.greet()</code></pre>

<p>
Methods help keep related data and behaviour together, which is one of the main goals of OOP.
</p>

```python
class Student:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, I am {self.name}.")


s1 = Student("Pete")
s1.greet() # This will print "Hello, I am Pete."
```

```python
class Marks:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def greet(self):
        print(f"My name is {self.name} and my exam mark is {self.marks}")


student1=Marks("Lifa", 90)
student1.greet()
```

```python
class Student:
    school = "NMU"  # Class attribute

    def __init__(self, name, marks):
        self.name = name    # Instance attribute
        self.marks = marks  # Instance attribute


s1 = Student("Lifa", 90)
s2 = Student("John", 75)

# Both the first student and the second student share the class attribute. 
# Instance attributes belong to one object and class attributes are share by the objects

print(s1.school)
print(s2.school)

print(s1.name, s1.marks)
print(s2.name, s2.marks)
```

<h2>Good Design from the Start</h2>

<p>
When learning OOP, many beginners create one large class that tries to do everything.
This is poor design.
</p>

<p>
A class should have one clear responsibility. This idea is known as the
<strong>Single Responsibility Principle (SRP)</strong>.
</p>

<p>
For example:
</p>

<ul>
<li>A <code>Student</code> class should manage student-related data.</li>
<li>A <code>Report</code> class could generate reports.</li>
<li>A <code>DatabaseManager</code> class could handle file or database storage.</li>
</ul>

<p>
Separating responsibilities makes programs easier to maintain, test, and extend.
</p>

<h2>Common Beginner Mistakes</h2>

<ul>
<li>Forgetting to include <code>self</code> in instance methods.</li>
<li>Using local variables instead of instance attributes.</li>
<li>Calling methods before creating an object.</li>
<li>Putting too much logic into one class.</li>
<li>Confusing class attributes with instance attributes.</li>
<li>Using unclear or badly named class names.</li>
</ul>

<p>
Readable code is an important professional skill. A programmer should be able to understand the purpose of
a class by reading its name and method names.
</p>
