### Encapsulation, Properties, Inheritance, and Polymorphism

<h2>Encapsulation in Python</h2>

<p>
Encapsulation is one of the core principles of object-oriented programming. It refers to the practice
of restricting direct access to an object's data and controlling how that data is modified.
</p>

<p>
In many languages, such as C#, encapsulation is enforced using access modifiers like <code>private</code>,
<code>protected</code>, and <code>public</code>. Python takes a different approach. Instead of enforcing
strict access control, Python relies on conventions and programmer discipline.
</p>

<p>
In Python, a single underscore (<code>_</code>) is used to indicate that an attribute is intended for
internal use, while a double underscore (<code>__</code>) triggers name mangling.
</p>

<p>
Encapsulation is important because it:
</p>

<ul>
<li>Protects data from unintended modification.</li>
<li>Allows validation before changing values.</li>
<li>Improves maintainability.</li>
<li>Supports cleaner design.</li>
</ul>

```python
class Student:

    def __init__(self, name, age):
        self.name = name          # public
        self._age = age           # protected (Python does not strictly enforce access control like C# does. 
                                  # Instead, it relies on naming conventions to signal intent to other developers.
```

<h2>Private Attributes and Name Mangling</h2>

<p>
When an attribute is prefixed with two underscores, Python performs a process known as
<strong>name mangling</strong>. This changes the internal name of the attribute to reduce the risk of
accidental access.
</p>

<p>
This does not make the attribute truly private, but it signals that it should not be accessed directly.
</p>

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.__age = age   # private

s = Student("John", 20)

# This will cause an error:
# print(s.__age)

# This still works (not recommended):
print(s._Student__age)
```

<h2>Why Encapsulation Matters</h2>

<p>
Without encapsulation, data can be changed freely, which may lead to incorrect or invalid program states.
</p>

<p>
For example, a student's age should not be negative. Encapsulation allows us to validate data before it is updated.
</p>

<h2>Using Properties in Python</h2>

<p>
Python provides a powerful and elegant way to implement encapsulation using <strong>properties</strong>.
</p>

<p>
Properties allow us to control access to attributes while still using simple syntax. Instead of calling
methods like <code>get_age()</code> and <code>set_age()</code>, we can use attribute-style access.
</p>

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative.")
        self._age = value
```

```python
s = Student("Mary", 21)

print(s.age)   # getter

s.age = 25     # setter

print(s.age)   # getter

# s.age = -5  # This will raise a ValueError because age cannot be negative.
```

<h2>Inheritance</h2>

<p>
Inheritance allows one class to reuse the attributes and methods of another class.
</p>

<p>
The class being inherited from is called the <strong>base class</strong> (or parent class), and the
class that inherits is called the <strong>derived class</strong> (or child class).
</p>

<p>
Inheritance promotes code reuse and logical organisation of related classes.
</p>

```python
class Person:

    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")


class Student(Person):

    def __init__(self, name, student_number):
        super().__init__(name)
        self.student_number = student_number
```

<h2>The <code>super()</code> Function</h2>

<p>
The <code>super()</code> function is used to call methods from the parent class.
</p>

<p>
This ensures that the parent class is properly initialised and avoids code duplication.
</p>

```python
s = Student("Thabo", "2026002")
s.display() # Provided by the Person class, not Student.
```

<h2>Method Overriding</h2>

<p>
A derived class can provide its own implementation of a method that already exists in the base class.
This is called <strong>method overriding</strong>.
</p>

<p>
Overriding allows a child class to customise behaviour while still maintaining a common interface.
</p>

```python
class Student(Person):

    def __init__(self, name, student_number):
        super().__init__(name)
        self.student_number = student_number

    def display(self):
        print(f"Name: {self.name}, Student Number: {self.student_number}")
```

```python
s = Student("Thabo", "2026002")
s.display() # Student's display method overrides Person's display method.
```

<h2>Polymorphism</h2>

<p>
Polymorphism means "many forms". In object-oriented programming, it allows different objects to respond
to the same method call in different ways.
</p>

<p>
This is one of the most powerful features of OOP because it allows us to write flexible and reusable code.
</p>

```python
class Lecturer(Person):

    def display(self):
        print(f"Lecturer Name: {self.name}")


people = [
    Student("John", "2026001"),
    Lecturer("Dr Smith")
]

for p in people:
    p.display()
```

<h2>Why Polymorphism Is Important</h2>

<p>
Polymorphism allows us to treat different types of objects in a uniform way.
</p>

<p>
Instead of writing separate code for each type, we can write general code that works with any object
that implements the required method.
</p>

<p>
This greatly simplifies program design and improves scalability.
</p>

<h2>Design Considerations</h2>

<p>
As systems grow, it is important to apply good design principles.
</p>

<ul>
<li>Use encapsulation to protect data.</li>
<li>Use inheritance only when there is a clear "is-a" relationship.</li>
<li>Use method overriding to customise behaviour.</li>
<li>Use polymorphism to simplify code.</li>
</ul>

<p>
Avoid using inheritance where it is not appropriate. In some cases, composition is a better design choice.
We will explore this in later lectures.
</p>

<h2>Common Mistakes</h2>

<ul>
<li>Accessing private attributes directly.</li>
<li>Forgetting to call <code>super()</code> in constructors.</li>
<li>Using inheritance where there is no logical relationship.</li>
<li>Overcomplicating class hierarchies.</li>
<li>Not validating data in setters.</li>
</ul>
